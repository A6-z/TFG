```python
import numpy as np
from pathlib import Path
from sklearn.decomposition import PCA
```

```python
embedding = Path("/gpfs/projects/bsc14/mouhida/TFG/embeddings")
model_emb = next(embedding.iterdir(), None)
embs_path = next(model_emb.iterdir(), None)

print(embs_path)
```

```python
data = np.load(embs_path)
print(data.files)
```

```python
names_test = data['test_filenames']
print(names_test[0])
```

```python
array1 = data["train_embeddings"]
print(array1.shape)
```

```python
array2 = data['test_embeddings']
arrays = np.concat((array1, array2), axis = 0)
print(arrays.shape)
```

```python
idx = np.random.choice(len(arrays), size = (20000, 2))  # choose 20k random pairs
idx[:, 1]
```

```python
pca = PCA()
pca.fit(arrays)  # shape (N, d)
cumvar = np.cumsum(pca.explained_variance_ratio_)
eff_dim_90 = np.searchsorted(cumvar, 0.90) + 1
```

```python
print(eff_dim_90)
```
